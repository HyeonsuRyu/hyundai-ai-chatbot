```python
from langchain_core.runnables import RunnableLambda


r_add = RunnableLambda(lambda x: x+1)
r_double = RunnableLambda(lambda x: x*2)

chain_a = r_add | r_double
chain_b = r_double | r_add

print(chain_a.invoke(3))
print(chain_b.invoke(3))
print(type(chain_a))
```

```python
from langchain_core.runnables import RunnableLambda, RunnableParallel, RunnablePassthrough

analyzer = RunnableParallel(
    original = RunnablePassthrough(),
    upper = RunnableLambda(lambda x: x.upper()),
    length = RunnableLambda(lambda x: len(x)),
)

print(analyzer.invoke("hello"))
```

```python
# 환경변수 로드
from dotenv import load_dotenv
load_dotenv()

# openai 클라이언트 생성
from openai import OpenAI
client = OpenAI()

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 프롬프트, 모델, 파서 생성 및 체인에 연결
prompt = ChatPromptTemplate.from_template("{topic}에 대해서 친절한 초등학교 선생님처럼 설명해줘")
model = ChatOpenAI(model = "gpt-4o-mini")
parser = StrOutputParser()

chain = prompt | model | parser

print(chain.invoke({"topic": "인공지능"}))
```

```python
results = chain.batch([
    {"topic": "블럭체인"},
    {"topic": "클라우드"},
])

for result in results:
    print("-", result)
    print("="*30)

for chunk in chain.stream({"topic": "양자컴퓨터"}):
    print(chunk, end="", flush=True)
```

```python
from langchain_core.runnables import RunnableLambda

r_strip = RunnableLambda(lambda x: x.strip())
r_upper = RunnableLambda(lambda x: x.upper())
r_exclamation = RunnableLambda(lambda x: x+"!")

chain = r_strip | r_upper | r_exclamation

print(chain.invoke(" hello wolrd   "))
```

```python
from langchain_core.runnables import RunnableParallel, RunnableLambda

runner = RunnableParallel(
    sum = RunnableLambda(lambda x: sum(x)),
    avg = RunnableLambda(lambda x: sum(x)/len(x)),
    max = RunnableLambda(lambda x: max(x)),
)

print(runner.invoke([10, 20, 30]))
```

```python
from dotenv import load_dotenv
load_dotenv()

from openai import OpenAI
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

client = OpenAI()

# 프롬프트, 모델, 파서 생성 및 체인에 연결
prompt1 = ChatPromptTemplate.from_template("{topic}에 대해서 한 줄로 설명해줘")
prompt2 = ChatPromptTemplate.from_template("{topic}에 대해서 관련 키워드 3개를 만들고, 부연 설명 없이 키워드와 뜻만 딕셔너리로 돌려줘")
model = ChatOpenAI(model = "gpt-4o-mini", temperature=0.0, max_completion_tokens=50)
parser = StrOutputParser()

chain1 = prompt1 | model | parser
chain2 = prompt2 | model | parser

runner = RunnableParallel(
    summary = chain1,
    keywords = chain2
)
for item in runner.invoke({"topic": "인공지능"}).items():
    print(item)
```
